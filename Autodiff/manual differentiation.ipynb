```python
import sympy as sp
```

```python
x = sp.symbols('x')
f = x**3 + 2*x**2 - 5*x + 1
print("f(x) =")
f
```

```python
df = sp.diff(f, x)
print("f'(x) =")
print(df)
```

```python
x_val = 2
fn = sp.lambdify(x, f)
df_fn = sp.lambdify(x, df)
print(f"f(x={x_val}) =",fn(x_val))
print(f"f'(x={x_val}) =",df_fn(x_val))
```
