```python
import tensorflow as tf
```

### Constant Tensors

```python
tf.ones(shape=(3, 3))
```

```python
tf.zeros(shape=(3, 3))
```

```python
tf.constant([1, 2, 3], dtype="float32")
```

```python
tf.constant([[1, 2, 3], [4, 5, 6], [7, 8, 9]], dtype="float32")
```

```python
a = tf.constant([[1, 2, 3], [4, 5, 6], [7, 8, 9]], dtype="float32")
# a[0][0] = 1 #Will throw error
```

### Random Tensors

```python
tf.random.normal(shape=(3, 3), mean=0., stddev=1.)
```

```python
tf.random.uniform(shape=(3, 3), minval=0., maxval=1.)
```

```python
a = tf.random.normal(shape=(3, 3), mean=0., stddev=1.)
a
```

### Variable tensor

```python
tf.Variable(initial_value=tf.random.normal(shape=(3, 3)))
```

```python
M = tf.Variable(initial_value=tf.random.normal(shape=(3, 3)))
M
```

```python
# M[0][0] = 0
# Will throw error
```

```python
# M[0, 0] = 0
# Will throw error
```

```python
M[0, 0].assign(0) # Assign value
M
```

```python
M[1, 1].assign(M[1, 1] + 100 ) # Add/Sub to element
M
```

```python
M.assign_add(tf.ones((3, 3)))
M
```
