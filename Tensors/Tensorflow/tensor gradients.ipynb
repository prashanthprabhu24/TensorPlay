```python
import tensorflow as tf
```

## Gradient of Rank 0 (Calculus)

### First Order Derivative

```python
input_var = tf.Variable(initial_value=3.0)
input_var
```

```python
with tf.GradientTape() as tape:
    result = tf.square(input_var) # x^2 (x=3) , dx = 2*x (2x = 6)
gradient = tape.gradient(result, input_var)
gradient
```

```python
pi = tf.constant(3.14)
pi
```

```python
with tf.GradientTape() as tape:
    tape.watch(pi)
    result = tf.square(pi) # x^2 (x=3.14) , dx = 2*x (2x = 6.28)
gradient = tape.gradient(result, pi)
gradient
```

### Second Order Derivative

```python
time = tf.Variable(0.)
time
```

```python
with tf.GradientTape() as outer_tape:
    with tf.GradientTape() as inner_tape:
        position = 4.9 * time**2  # c * x**2 (x=0), dx = 2cx (2cx = 0), dx^2 = 2c (2c = 9.8)
    speed = inner_tape.gradient(position, time)
    print(speed)
acceleration = outer_tape.gradient(speed, time)
print(acceleration)
```

## Gradient of Rank 1 (Vector Calculus)

```python
input_var = tf.Variable(initial_value=[1., 2., 3.])
input_var
```

```python
with tf.GradientTape() as tape:
    result = tf.square(input_var) # x^2, dx = 2*x
gradient = tape.gradient(result, input_var)
gradient
```

```python
time = tf.Variable([0., 1., 2., 3.])
with tf.GradientTape() as outer_tape:
    with tf.GradientTape() as inner_tape:
        position = 4.9 * time**2  # c * x**2, dx = 2cx, dx^2 = 2c
    speed = inner_tape.gradient(position, time)
    print(speed)
acceleration = outer_tape.gradient(speed, time)
print(acceleration)
```

## Gradient of Rank 2+ (Tensor Calculus)

```python
input_var = tf.Variable(initial_value=[[1., 2., 3.], [4., 5., 6.], [7., 8., 9.]])
input_var
```

```python
with tf.GradientTape() as tape:
    result = tf.square(input_var) # x^2, dx = 2*x
gradient = tape.gradient(result, input_var)
gradient
```

```python
time = tf.Variable([[0., 1., 2.], [3., 4., 5.], [6., 7., 8.]])
with tf.GradientTape() as outer_tape:
    with tf.GradientTape() as inner_tape:
        position = 4.9 * time**2  # c * x**2, dx = 2cx, dx^2 = 2c
    speed = inner_tape.gradient(position, time)
    print(speed)
acceleration = outer_tape.gradient(speed, time)
acceleration
```
