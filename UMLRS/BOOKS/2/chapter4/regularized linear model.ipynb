### Early Stopping

```python
rng = np.random.default_rng(seed=42)
m = 200
X = 6 * rng.random((m, 1))- 3
X
```

```python
theta0 = 2
theta1 = 1
theta2 = 0.5
noise = rng.standard_normal((m, 1))
y = theta2 * X ** 2 + theta1 * X + theta0 + noise
y
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size=0.2)
```

```python
preprocessing = make_pipeline(PolynomialFeatures(degree=90, include_bias=False), StandardScaler())
X_train_prep = preprocessing.fit_transform(X_train)
X_valid_prep = preprocessing.transform(X_valid)
```

```python
history_train = []
history_val = []
n_epochs = 200
best_valid_rmse = float('inf')
best_model_epoch = None
best_model = None
sgd_reg = SGDRegressor(penalty=None, eta0=0.002, random_state=42)

for epoch in range(n_epochs):
    sgd_reg.partial_fit(X_train_prep, y_train.ravel())
    y_train_predict = sgd_reg.predict(X_train_prep)
    y_valid_predict = sgd_reg.predict(X_valid_prep)
    training_error = root_mean_squared_error(y_train, y_train_predict)
    val_error = root_mean_squared_error(y_valid, y_valid_predict)
    history_train.append(training_error)
    history_val.append(val_error)
    if val_error < best_valid_rmse:
        best_valid_rmse = val_error
        best_model = deepcopy(sgd_reg)
        best_model_epoch = epoch

plt.figure(figsize=(12, 8))
plt.plot(list(range(n_epochs)), history_train, label="Training Error")
plt.plot(list(range(n_epochs)), history_val, label="Validation Error")
plt.scatter(best_model_epoch, history_val[best_model_epoch], color="r", label="Best Model")
plt.grid()
plt.legend()
plt.show()
```
