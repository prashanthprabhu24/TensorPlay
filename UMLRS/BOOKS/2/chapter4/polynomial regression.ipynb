```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
rng = np.random.default_rng(seed=42)
m = 200
X = 6 * rng.random((m, 1))- 3
X
```

```python
theta0 = 2
theta1 = 1
theta2 = 0.5
noise = rng.standard_normal((m, 1))
y = theta2 * X ** 2 + theta1 * X + theta0 + noise
y
```

```python
plt.scatter(X, y)
plt.grid()
plt.show()
```

```python
poly_features = PolynomialFeatures(degree=2, include_bias=False)
poly_features
```

```python
X_poly = poly_features.fit_transform(X)
X_poly
```

```python
lin_reg = LinearRegression()
lin_reg.fit(X_poly, y)
lin_reg.intercept_, lin_reg.coef_
```

```python
theta0 = lin_reg.intercept_[0]
theta1 = lin_reg.coef_[0][0]
theta2 = lin_reg.coef_[0][1]
```

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(
    X_poly[:, 0],
    X_poly[:, 1],
    y.ravel(),
    label="Data"
)

x1 = np.linspace(X_poly[:, 0].min(), X_poly[:, 0].max(), 30)
x2 = np.linspace(X_poly[:, 1].min(), X_poly[:, 1].max(), 30)
X1, X2 = np.meshgrid(x1, x2)
Y_plane = theta0 + theta1 * X1 + theta2 * X2
ax.plot_surface(X1, X2, Y_plane, alpha=0.5)
ax.set_xlabel("x")
ax.set_ylabel("x²")
ax.set_zlabel("y")
plt.show()
```

```python
def f(theta0, theta1, theta2, X):
    return theta2 * X ** 2 + theta1 * X + theta0

plt.scatter(X, y)
plt.plot(np.linspace(-3, 3, 200), f(theta0, theta1, theta2, np.linspace(-3, 3, 200)), "r", linewidth=2, label="Predictions")
plt.xlabel("X")
plt.ylabel("y")
plt.grid()
plt.legend()
plt.show()
```

```python
poly_features1d = PolynomialFeatures(degree=1, include_bias=False)
poly_features2d = PolynomialFeatures(degree=2, include_bias=False)
poly_features10d = PolynomialFeatures(degree=10, include_bias=False)
poly_features100d = PolynomialFeatures(degree=20, include_bias=False)

X_poly1 = poly_features1d.fit_transform(X)
X_poly2 = poly_features2d.fit_transform(X)
X_poly10 = poly_features10d.fit_transform(X)
X_poly100 = poly_features100d.fit_transform(X)


lin_reg1 = LinearRegression()
lin_reg2 = LinearRegression()
lin_reg10 = LinearRegression()
lin_reg100 = LinearRegression()

lin_reg1.fit(X_poly1, y)
lin_reg2.fit(X_poly2, y)
lin_reg10.fit(X_poly10, y)
lin_reg100.fit(X_poly100, y)

theta0_1, theta_1  = lin_reg1.intercept_[0], lin_reg1.coef_[0]
theta0_2, theta_2  = lin_reg2.intercept_[0], lin_reg2.coef_[0]
theta0_10, theta_10  = lin_reg10.intercept_[0], lin_reg10.coef_[0]
theta0_100, theta_100  = lin_reg100.intercept_[0], lin_reg100.coef_[0]
def f(theta0, thetax, X):
    val = theta0
    for i in range(0, len(thetax)):
        val += (X**(i+1)) * thetax[i]
    return val
    # return theta2 * X ** 2 + theta1 * X + theta0

plt.scatter(X, y)
plt.plot(np.linspace(-3, 3, 200), f(theta0_1, theta_1, np.linspace(-3, 3, 200)), "r", linewidth=2, label="Degree - 1")
plt.plot(np.linspace(-3, 3, 200), f(theta0_2, theta_2, np.linspace(-3, 3, 200)), "g", linewidth=2, label="Degree - 2")
plt.plot(np.linspace(-3, 3, 200), f(theta0_10, theta_10, np.linspace(-3, 3, 200)), "b", linewidth=2, label="Degree - 10")
plt.plot(np.linspace(-3, 3, 200), f(theta0_100, theta_100, np.linspace(-3, 3, 200)), "pink", linewidth=2, label="Degree - 100")
plt.xlabel("X")
plt.ylabel("y")
plt.grid()
#plt.legend()
plt.show()
```
