```python
import numpy as np
```

```python
def step_function(z, threshold=0):
    return (z >= threshold).astype(int)
```

```python
def initialize_weights(n_inputs, n_outputs):
    weights = np.random.randn(n_outputs, n_inputs)
    bias = np.random.randn(n_outputs)
    return weights, bias
```

```python
def forward(X, weights, bias):
    z = X @ weights.T + bias
    y_pred = step_function(z)
    return y_pred
```

```python
def score(X, y, weights, bias):
    predictions = predict(X, weights, bias)
    y = y.reshape(-1, predictions.shape[1])
    return np.mean(predictions == y)
```

```python
def predict(X, weights, biases):
    z = X @ weights.T + biases
    return step_function(z)
```

```python
def train(X, y, epochs=100, learning_rate=1):
    n_inputs = X.shape[1]
    n_outputs = y.shape[1] if len(y.shape) == 2 else 1
    weights, bias = initialize_weights(n_inputs, n_outputs)
    for epoch in range(epochs):
        for x, target in zip(X, y):
            prediction = forward(x.reshape(1, -1), weights, bias)[0]
            error = target - prediction
            weights += (learning_rate * np.outer(error, x))
            bias += learning_rate * error
    return weights, bias
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]]) # AND
y = np.array([0, 0, 0, 1])

weights, bias = train(X, y, epochs=10)

print("\nWeights:", weights)
print("Bias:", bias)

print("\nPredictions:")
print(predict(X, weights, bias))

print("\nAccuracy:", score(X, y, weights, bias))
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]]) # Flips
y = np.array([[1, 1], [1, 0], [0, 1], [0, 0]])

weights, bias = train(X, y, epochs=10)

print("\nWeights:", weights)
print("Bias:", bias)

print("\nPredictions:")
print(predict(X, weights, bias))

print("\nAccuracy:", score(X, y, weights, bias))
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]]) # OR
y = np.array([0, 1, 1, 1])

weights, bias = train(X, y, epochs=5)

print("\nWeights:", weights)
print("Bias:", bias)

print("\nPredictions:")
print(predict(X, weights, bias))

print("\nAccuracy:", score(X, y, weights, bias))
```
