```python
import numpy as np
```

```python
def activation(z):
    return np.where(z >= 0, 1, 0)
```

```python
def predict(X, weights, bias):
    z = X @ weights + bias
    return activation(z)
```

```python
def score(X, y, weights, bias):
    predictions = predict(X, weights, bias)
    return np.mean(predictions == y)
```

```python
def fit(X, y, epochs=1, learning_rate=1):
    n_features = X.shape[1]
    weights = np.zeros(n_features)
    bias = 0.0
    for epoch in range(epochs):
        errors = 0
        for xi, yi in zip(X, y):
            z = np.dot(xi, weights) + bias
            y_pred = activation(z)
            error = yi - y_pred
            weights += learning_rate * error * xi
            bias += learning_rate * error
            if error != 0:
                errors += 1
        print(f"Epoch: {epoch + 1} | ", f"Errors: {errors}")
        if errors == 0:
            break
    return weights, bias
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]]) # AND
y = np.array([0, 0, 0, 1])

weights, bias = fit(X, y, epochs=10)

print("\nWeights:", weights)
print("Bias:", bias)

print("\nPredictions:")
print(predict(X, weights, bias))

print("\nAccuracy:", score(X, y, weights, bias))
```
