```python
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import export_graphviz
from graphviz import Source
import numpy as np
from sklearn.model_selection import train_test_split
```

### Load Datasets

```python
X, y = make_moons(n_samples=150, noise=0.2, random_state=42)
X_moons, X_test, y_moons, y_test = train_test_split(X, y, test_size=0.2)
```

### Boundry Analysis

```python
def decision_boundary(tree, X, Y, X_test, y_test):
    t = tree.tree_
    x_min, x_max = X[:, 0].min(), X[:, 0].max()
    y_min, y_max = X[:, 1].min(), X[:, 1].max()
    plt.figure(figsize=(8, 6))
    def traverse(node, xlim, ylim):
        feature = t.feature[node]
        threshold = t.threshold[node]
        if feature == -2:
            predicted_class = np.argmax(t.value[node][0])
            color = "yellow" if predicted_class == 0 else "orange"
            plt.fill([xlim[0], xlim[1], xlim[1], xlim[0]], [ylim[0], ylim[0], ylim[1], ylim[1]], color=color, alpha=0.2)
            return
        if feature == 0:
            traverse(t.children_left[node], (xlim[0], threshold), ylim)
            traverse(t.children_right[node], (threshold, xlim[1]), ylim)
        elif feature == 1:
            traverse(t.children_left[node], xlim, (ylim[0], threshold))
            traverse(t.children_right[node], xlim, (threshold, ylim[1]))
    traverse(0, (x_min, x_max), (y_min, y_max))
    plt.scatter(X[Y == 0, 0], X[Y == 0, 1], color="blue", edgecolor="black", label="Class 0")
    plt.scatter(X[Y == 1, 0], X[Y == 1, 1], color="orange", edgecolor="black", label="Class 1")
    plt.scatter(X_test[y_test == 0, 0], X_test[y_test == 0, 1], color="green", edgecolor="white", label="Class test 0")
    plt.scatter(X_test[y_test == 1, 0], X_test[y_test == 1, 1], color="yellow", edgecolor="white", label="Class test 1")
    plt.xlabel("X1")
    plt.ylabel("X2")
    plt.title("Decision Boundary")
    plt.legend()
    plt.grid()
    plt.show()
```

```python
def fit_boundary(depth=1):
    tree_clfx = DecisionTreeClassifier(max_depth=depth)
    tree_clfx.fit(X_moons, y_moons)
    decision_boundary(tree_clfx, X_moons, y_moons, X_test, y_test)
    print("Score :", tree_clfx.score(X_test, y_test) )
    if depth <= 3:
        dot = export_graphviz(tree_clfx, out_file=None, rounded=True, filled=True)
        return Source(dot)
```
