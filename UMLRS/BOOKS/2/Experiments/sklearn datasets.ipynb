```python
roll = datasets.make_swiss_roll(100000)
roll
```

```python
# Plot swiss roll
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(roll[0][:, 0], roll[0][:, 1], roll[0][:, 2], c=roll[1], cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Swiss Roll Dataset')
plt.show()
```
