```python
import random
import matplotlib.pyplot as plt
```

```python
def get_coin(bias=0.5):
    coin = [0, 1], [1-bias, bias] # [tail, head], [Prob of Tail, Prob of Head]
    return coin
```

```python
def toss_coin(coin):
    face = random.choices(coin[0], coin[1])
    return face[0]
```

```python
coin = get_coin()
face  = toss_coin(coin)
face
```

```python
def toss_n(coin, n):
    no_of_heads = 0
    toss_graph = []
    for i in range(1, n):
        face = toss_coin(coin)
        if face:
            no_of_heads += 1
        toss_graph.append(no_of_heads/i)
    return toss_graph
```

```python
def toss_m_coins(m=1, bias = 0.5, n=1000):
    final_mean_last_toss_face = 0
    last_toss_face = []
    for i in range(m):
        coin = get_coin(bias)
        coin_graph = toss_n(coin, n)
        plt.plot(list(range(1, n)), coin_graph)
        last_toss_face.append(coin_graph[-1])
    final_mean_last_toss_face = sum(last_toss_face)/m
    return final_mean_last_toss_face
```

```python
toss_m_coins(10, bias=0.51)
```

```python
toss_m_coins(10, bias=0.51, n=1000000)
```

```python
toss_m_coins(10000, bias=0.51, n=1000)
```
