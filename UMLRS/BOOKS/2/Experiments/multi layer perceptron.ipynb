```python
import numpy as np
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def sigmoid_derivative(z):
    s = sigmoid(z)
    return s * (1 - s)
```

```python
def initialize_network(architecture):
    weights = []
    biases = []
    for i in range(len(architecture) - 1):
        n_inputs = architecture[i]
        n_neurons = architecture[i + 1]
        W = np.random.randn(n_neurons, n_inputs) * 0.1
        b = np.zeros(n_neurons)
        weights.append(W)
        biases.append(b)
    return weights, biases
```

```python
def forward(x, weights, biases):
    activations = [x]
    z_values = []
    a = x
    for W, b in zip(weights, biases):
        z = W @ a + b
        a = sigmoid(z)
        z_values.append(z)
        activations.append(a)
    return activations, z_values
```

```python
def train(X, y, architecture, learning_rate=0.1, epochs=1000):
    weights, biases = initialize_network(architecture)
    for epoch in range(epochs):
        total_error = 0
        for x, target in zip(X, y):
            activations, z_values = forward(x, weights, biases)
            prediction = activations[-1]
            error = target - prediction
            total_error += np.sum(error ** 2)
            deltas = [None] * len(weights)
            deltas[-1] = (error * sigmoid_derivative(z_values[-1]))
            for layer in range(len(weights) - 2, -1, -1):
                deltas[layer] = (weights[layer + 1].T @ deltas[layer + 1] * sigmoid_derivative( z_values[layer]))
            for layer in range(len(weights)):
                weights[layer] += (learning_rate * np.outer(deltas[layer], activations[layer]))
                biases[layer] += (learning_rate * deltas[layer])
        if epoch % 100 == 0:
            print(f"Epoch {epoch:5d} | " f"Error: {total_error:.6f}")
    return weights, biases
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
y = np.array([[0], [1], [1], [0]], dtype=float)
architecture = [2, 3, 1]
train(X, y, architecture)
```
