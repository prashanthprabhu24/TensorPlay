```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.tree import export_graphviz
from graphviz import Source
```

### Load Dataset

```python
rng = np.random.default_rng(seed=42)
X = rng.random((250, 1))- 0.5
X
```

```python
noise = 0.025 * rng.standard_normal((250, 1))
y = X ** 2 + noise
y
```

```python
X_quad, X_test, y_quad, y_test = train_test_split(X, y, test_size=0.2)
```

```python
plt.scatter(X_quad, y_quad)
```

### Model

```python
tree_reg = DecisionTreeRegressor(max_depth=2, random_state=42)
tree_reg.fit(X_quad, y_quad)
dot = export_graphviz(tree_reg, out_file=None, rounded=True, filled=True)
Source(dot)
```

```python
tree_reg.score(X_test, y_test)
```
