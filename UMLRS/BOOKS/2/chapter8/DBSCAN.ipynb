```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
```

### Load data

```python
X, y = make_moons(n_samples=1000, noise=0.05, random_state=42)
y.shape
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="prism")
plt.show()
```

### Model

```python
dbscan = DBSCAN(eps=0.05, min_samples=5)
dbscan.fit(X)
```

```python
len(dbscan.labels_), dbscan.labels_
```

```python
len(dbscan.core_sample_indices_), dbscan.core_sample_indices_
```

```python
dbscan.components_
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="prism")
plt.scatter(dbscan.components_[:, 0], dbscan.components_[:, 1], c=dbscan.labels_[dbscan.core_sample_indices_])
plt.scatter(X[dbscan.labels_==-1][:, 0], X[dbscan.labels_==-1][:, 1], marker="x", color="white")
plt.show()
```

```python
dbscan = DBSCAN(eps=0.05, min_samples=5)
dbscan.fit(X)
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.scatter(dbscan.components_[:, 0], dbscan.components_[:, 1], c=dbscan.labels_[dbscan.core_sample_indices_], cmap="prism")
plt.scatter(X[dbscan.labels_==-1][:, 0], X[dbscan.labels_==-1][:, 1], marker="x", color="white")
plt.show()
```
