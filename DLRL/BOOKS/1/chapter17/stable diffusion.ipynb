### Load Model

```python
height, width = 512, 512
stable_diffusion_model = keras_hub.models.TextToImage.from_preset("stable_diffusion_3_medium", image_shape=(height, width, 3), dtype="float16")
stable_diffusion_model.summary()
```

### Inference

```python
prompt = "A NASA astraunaut riding an origami elephant in New York City"
stable_diffusion_model.generate(prompt)
```

```python
stable_diffusion_model.generate({"prompts": prompt, "negative_prompts": "blue color"})
```

```python
def display(images):
    return Image.fromarray(np.concatenate(images, axis=1))
display([stable_diffusion_model.generate(prompt, num_steps=x) for x in [5, 10, 15, 20, 25]])
```
