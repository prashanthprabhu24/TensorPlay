```python
import numpy as np
import matplotlib.pyplot as plt
import keras
```

```python
(train_images, train_labels), (test_images, test_labels) = keras.datasets.mnist.load_data()
train_images.shape, train_labels.shape, test_images.shape, test_labels.shape
```

```python
train_images = train_images.reshape((60000, 28 * 28))
train_images = train_images.astype("float32") / 255
test_images = test_images.reshape((10000, 28 * 28))
test_images = test_images.astype("float32") / 255

train_images.shape, test_images.shape
```

### Learning Rate Tuning

```python
def get_model(optimizer=keras.optimizers.RMSprop(learning_rate=1.0)):
    model = keras.Sequential(
        [
            keras.layers.Dense(512, activation="relu"),
            keras.layers.Dense(10, activation="softmax"),
        ]
    )
    model.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )
    return model
```

```python
modele10 = get_model(optimizer=keras.optimizers.RMSprop(learning_rate=1e-10))
historye10 = modele10.fit(
    train_images,
    train_labels,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=0
)
```

```python
modele5 = get_model(optimizer=keras.optimizers.RMSprop(learning_rate=1e-5))
historye5 = modele5.fit(
    train_images,
    train_labels,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=0
)
```

```python
modele3 = get_model(optimizer=keras.optimizers.RMSprop(learning_rate=1e-3))
historye3 = modele3.fit(
    train_images,
    train_labels,
    epochs=30,
    batch_size=128,
    validation_split=0.2,
    verbose=0
)
```

```python
modele2 = get_model(optimizer=keras.optimizers.RMSprop(learning_rate=1e-2))
historye2 = modele2.fit(
    train_images,
    train_labels,
    epochs=30,
    batch_size=128,
    validation_split=0.2,
    verbose=0
)
```

```python
val_acc_e10 = historye10.history["val_accuracy"]
val_acc_e5 = historye5.history["val_accuracy"]
val_acc_e3 = historye3.history["val_accuracy"]
val_acc_e2 = historye2.history["val_accuracy"]
val_acc_e1 = historye1.history["val_accuracy"]
val_acc_1 = history1.history["val_accuracy"]
val_acc_3 = history3.history["val_accuracy"]
epochs = range(1, 31)
#plt.plot(epochs,val_acc_e10,label="e10")
#plt.plot(epochs,val_acc_e5,label="e5")
plt.plot(epochs,val_acc_e3,label="e3")
plt.plot(epochs,val_acc_e2,label="e2")
plt.plot(epochs,val_acc_e1,label="e1")
#plt.plot(epochs,val_acc_1,label="1")
#plt.plot(epochs,val_acc_3,label="3")

plt.title("Effect of Different LR Value")
plt.xlabel("Epochs")
plt.xticks(epochs)
plt.ylabel("Accuracy")
plt.legend()
plt.show()
```

### Model Size Tuning

```python
model = keras.Sequential([keras.layers.Dense(10, activation="softmax")])
model.compile(
    optimizer="rmsprop",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
history_small_model = model.fit(
train_images, train_labels, epochs=20, batch_size=128, validation_split=0.2
)
```

```python
val_loss = history_small_model.history["val_loss"]
epochs = range(1, 21)
plt.plot(epochs, val_loss, "b-", label="Validation loss")
plt.title("Validation loss for a model with insufficient capacity")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

```python
model = keras.Sequential(
[
        keras.layers.Dense(128, activation="relu"),
        keras.layers.Dense(128, activation="relu"),
        keras.layers.Dense(10, activation="softmax"),
]
)
model.compile(
    optimizer="rmsprop",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
history_large_model = model.fit(
    train_images,
    train_labels,
    epochs=20,
    batch_size=128,
    validation_split=0.2,
)
```

```python
val_loss = history_large_model.history["val_loss"]
epochs = range(1, 21)
plt.plot(epochs, val_loss, "b-", label="Validation loss")
plt.title("Validation loss for a model with sufficient capacity")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

```python
model = keras.Sequential(
[
        keras.layers.Dense(2048, activation="relu"),
        keras.layers.Dense(2048, activation="relu"),
        keras.layers.Dense(2048, activation="relu"),
        keras.layers.Dense(10, activation="softmax"),
]
)
model.compile(
    optimizer="rmsprop",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
history_very_large_model = model.fit(
    train_images,
    train_labels,
    epochs=20,
    batch_size=32,
    validation_split=0.2,
)
```
