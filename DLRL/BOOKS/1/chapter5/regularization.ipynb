```python
import numpy as np
import matplotlib.pyplot as plt
import keras
```

### Load Dataset

```python
(train_data, train_labels), _ = keras.datasets.imdb.load_data(num_words=10000)
train_data.shape, train_labels.shape
```

```python
train_data[0]
```

```python
train_labels
```

```python
def vectorize_sequences(sequences, dimension=10000):
    results = np.zeros((len(sequences), dimension))
    for i, sequence in enumerate(sequences):
        results[i, sequence] = 1.0
    return results
train_data = vectorize_sequences(train_data)
train_data[0]
```

### Modelling

```python
model = keras.Sequential(
    [
        keras.layers.Dense(16, activation="relu"),
        keras.layers.Dense(16, activation="relu"),
        keras.layers.Dense(1, activation="sigmoid"),
    ]
)
model.compile(
    optimizer="rmsprop",
    loss="binary_crossentropy",
    metrics=["accuracy"],
)
history_original = model.fit(
    train_data,
    train_labels,
    epochs=20,
    batch_size=512,
    validation_split=0.4,
)
```

```python
model_small = keras.Sequential(
    [
        keras.layers.Dense(4, activation="relu"),
        keras.layers.Dense(4, activation="relu"),
        keras.layers.Dense(1, activation="sigmoid"),
    ]
)
model_small.compile(
    optimizer="rmsprop",
    loss="binary_crossentropy",
    metrics=["accuracy"],
)
history_small = model_small.fit(
    train_data,
    train_labels,
    epochs=20,
    batch_size=512,
    validation_split=0.4,
)
```

```python
model_big = keras.Sequential(
    [
        keras.layers.Dense(512, activation="relu"),
        keras.layers.Dense(512, activation="relu"),
        keras.layers.Dense(1, activation="sigmoid"),
    ]
)
model_big.compile(
    optimizer="rmsprop",
    loss="binary_crossentropy",
    metrics=["accuracy"],
)
history_big = model_big.fit(
    train_data,
    train_labels,
    epochs=20,
    batch_size=512,
    validation_split=0.4,
)
```

```python
epochs = range(1, 21)
plt.plot(epochs, history_original.history["loss"], label="Loss")
plt.plot(epochs, history_small.history["loss"], label="Smaller model loss")
plt.plot(epochs, history_big.history["loss"], label="Big model loss")

plt.title("Training Loss for a model with different capacity")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()

plt.show()
```

```python
epochs = range(1, 21)
plt.plot(epochs, history_original.history["val_loss"], label="Validation loss")
plt.plot(epochs, history_small.history["val_loss"], label="Smaller model Validation loss")
plt.plot(epochs, history_big.history["val_loss"], label="Big model Validation loss")

plt.title("Validation loss for a model with different capacity")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()

plt.show()
```

### Regularization Type

```python
model_l1 = keras.Sequential(
    [
        keras.layers.Dense(512, activation="relu", kernel_regularizer=keras.regularizers.l1(0.002)),
        keras.layers.Dense(512, activation="relu", kernel_regularizer=keras.regularizers.l1(0.002)),
        keras.layers.Dense(1, activation="sigmoid"),
    ]
)
model_l1.compile(
    optimizer="rmsprop",
    loss="binary_crossentropy",
    metrics=["accuracy"],
)
history_l1 = model_l1.fit(
    train_data,
    train_labels,
    epochs=20,
    batch_size=512,
    validation_split=0.4,
)
```

```python
model_l2 = keras.Sequential(
    [
        keras.layers.Dense(512, activation="relu", kernel_regularizer=keras.regularizers.l2(0.002)),
        keras.layers.Dense(512, activation="relu", kernel_regularizer=keras.regularizers.l2(0.002)),
        keras.layers.Dense(1, activation="sigmoid"),
    ]
)
model_l2.compile(
    optimizer="rmsprop",
    loss="binary_crossentropy",
    metrics=["accuracy"],
)
history_l2 = model_l2.fit(
    train_data,
    train_labels,
    epochs=20,
    batch_size=512,
    validation_split=0.4,
)
```

```python
model_drop = keras.Sequential(
    [
        keras.layers.Dense(512, activation="relu"),
        keras.layers.Dropout(0.5),
        keras.layers.Dense(512, activation="relu"),
        keras.layers.Dropout(0.5),
        keras.layers.Dense(1, activation="sigmoid"),
    ]
)
model_drop.compile(
    optimizer="rmsprop",
    loss="binary_crossentropy",
    metrics=["accuracy"],
)
history_drop = model_drop.fit(
    train_data,
    train_labels,
    epochs=20,
    batch_size=512,
    validation_split=0.4,
)
```

```python
epochs = range(1, 21)
plt.plot(epochs, history_original.history["val_loss"], label="Validation loss")
plt.plot(epochs, history_big.history["val_loss"], label="Big model Validation loss")
#plt.plot(epochs, history_l1.history["val_loss"], label="Big model L1 Validation loss")
#plt.plot(epochs, history_l2.history["val_loss"], label="Big model L2 Validation loss")
plt.plot(epochs, history_drop.history["val_loss"], label="Big model DropOut Validation loss")

plt.title("Validation loss for a model with Regularization")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()

plt.show()
```
