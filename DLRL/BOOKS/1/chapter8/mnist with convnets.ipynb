```python
import keras
```

```python
# Data Loading and Preprocessing
mnist = keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train = X_train.reshape((60000, 28, 28, 1))
X_train = X_train.astype("float32") / 255
X_test = X_test.reshape((10000, 28, 28, 1))
X_test = X_test.astype("float32") / 255
```

```python
# ConvNet Model Arch
inputs = keras.Input(shape=(28, 28, 1))
layer1 = keras.layers.Conv2D(filters=64, kernel_size=3, activation="relu")(inputs)
layer1_out = keras.layers.MaxPooling2D(pool_size=2)(layer1)
layer2 = keras.layers.Conv2D(filters=128, kernel_size=3, activation="relu")(layer1_out)
layer2_out = keras.layers.MaxPooling2D(pool_size=2)(layer2)
layer3 = keras.layers.Conv2D(filters=256, kernel_size=3, activation="relu")(layer2_out)
layer_3_out = keras.layers.GlobalAveragePooling2D()(layer3)
outputs = keras.layers.Dense(10, activation="softmax")(layer_3_out)
model = keras.Model(inputs=inputs, outputs=outputs)
model.summary()
```

```python
# Model Training
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
model.fit(X_train, y_train, epochs=5, batch_size=64)
```

```python
# Model Evaluation
test_loss, test_acc = model.evaluate(X_test, y_test)
test_loss, test_acc
```

```python
# Model Training
model_no_max_pool.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
model_no_max_pool.fit(X_train, y_train, epochs=5, batch_size=64)
```

```python
# Model Evaluation
test_loss, test_acc = model_no_max_pool.evaluate(X_test, y_test)
test_loss, test_acc
```
