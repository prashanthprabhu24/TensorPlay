```python
import tensorflow as tf
import jax
from jax import numpy as jnp
import torch
```

### First Order Derivative

```python
initial_value = 3.0
```

```python
# Tensorflow
input_var = tf.Variable(initial_value=initial_value)
with tf.GradientTape() as tape:
    result = tf.square(input_var) # X^2
gradient = tape.gradient(result, input_var) # 2*X
result.numpy().item(), gradient.numpy().item()
```

```python
# Torch
input_var = torch.tensor(initial_value, requires_grad=True)
result = torch.square(input_var) # X^2
result.backward() # 2*X
gradient = input_var.grad
input_var.grad = None
result.item(), gradient.item()
```

```python
# JAX
input_var = jnp.array(initial_value)
def functionA(input_var):
    return jnp.square(input_var)

grad_functionA = jax.grad(functionA) # X^2, 2*X
result = functionA(input_var)
gradient = grad_functionA(input_var)
result.item(), gradient.item()
```

### Second Order Derivative

```python
initial_value = 0.4
```

```python
# Tensorflow
time = tf.Variable(initial_value)
with tf.GradientTape() as outer_tape:
    with tf.GradientTape() as inner_tape:
        position = 4.9 * time**2
    speed = inner_tape.gradient(position, time)
acceleration = outer_tape.gradient(speed, time)

print(f"Time:         {time.numpy().item():.4f}")
print(f"Position:     {position.numpy().item():.4f}")
print(f"Speed:        {speed.numpy().item():.4f}")
print(f"Acceleration: {acceleration.numpy().item():.4f}")
```

```python
# Torch
time = torch.tensor(initial_value, requires_grad=True)
position = 4.9 * time**2
position.backward(create_graph=True)
speed = time.grad
time.grad = None
speed.backward()
acceleration = time.grad
time.grade = None
print(f"Time:         {time.item():.4f}")
print(f"Position:     {position.item():.4f}")
print(f"Speed:        {speed.item():.4f}")
print(f"Acceleration: {acceleration.item():.4f}")
```

```python
# JAX
time = jnp.array(initial_value)

def get_position(time):
    return 4.9 * time**2

get_speed = jax.grad(get_position)
get_acceleration = jax.grad(get_speed)
position = get_position(time)
speed = get_speed(time)
acceleration = get_acceleration(time)

print(f"Time:         {time.item():.4f}")
print(f"Position:     {position.item():.4f}")
print(f"Speed:        {speed.item():.4f}")
print(f"Acceleration: {acceleration.item():.4f}")
```
