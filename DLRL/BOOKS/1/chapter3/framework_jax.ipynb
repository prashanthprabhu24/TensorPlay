```python
def functionA(input_var):
    return jnp.square(input_var)
```

```python
grad_functionA = jax.grad(functionA)
grad_functionA
```

```python
input_var = jnp.array(3.0)
input_var
```

```python
functionA(input_var)
```

```python
grad_functionA(input_var)
```
