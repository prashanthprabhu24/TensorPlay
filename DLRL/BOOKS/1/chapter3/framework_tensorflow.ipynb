```python
input_var = tf.Variable(initial_value=3.0)
with tf.GradientTape() as tape:
    result = tf.square(input_var)
gradient = tape.gradient(result, input_var)
result, gradient
```

```python
input_var = tf.Variable(initial_value=3.0)
input_var
```

```python
with tf.GradientTape() as tape:
    res1 = input_var*2
    res2 = res1**3
    res3 = res2**0.5
    result = res3
gradient = tape.gradient(result, input_var)
result, gradient
```

```python
input_const = tf.constant(3.0)
with tf.GradientTape() as tape:
    tape.watch(input_const)
    result = tf.square(input_const)
gradient = tape.gradient(result, input_const)
result, gradient
```

```python
time = tf.Variable(0.11)
with tf.GradientTape() as outer_tape:
    with tf.GradientTape() as inner_tape:
        position = 4.9 * time**2
    speed = inner_tape.gradient(position, time)
acceleration = outer_tape.gradient(speed, time)
time, position, speed, acceleration
```

```python
@tf.function  # Compiling function to make it run faster - graph mode (Faster)
def denseA(inputs, W, b):
    return tf.nn.relu(tf.matmul(inputs, W) + b)

```

```python
@tf.function(jit_compile=True) #  # Compiling function to make it run faster - XLA mode (Very fast)
def denseB(inputs, W, b):
    return tf.nn.relu(tf.matmul(inputs, W) + b)

```

```python
input_size=10000
output_size=10000

W = tf.keras.Variable(shape=(input_size, output_size), initializer="uniform")
b = tf.keras.Variable(shape=(output_size,), initializer="uniform")
inputs = tf.keras.Variable(shape=(input_size, input_size), initializer="uniform")
W.shape, b.shape, inputs.shape
```

```python
import time

start1 = time.time()
res1 = denseA(inputs, W, b)
end1 = time.time()
end1 - start1
```

```python
start2 = time.time()
res2 = denseB(inputs, W, b)
end2 = time.time()
end2 - start2
```
