```python
def dense(inputs, W, b):
    return torch.nn.relu(torch.matmul(inputs, W) + b)
```

```python
input_var = torch.tensor(3.0, requires_grad=True)
input_var
```

```python
result = torch.square(input_var)
result
```

```python
result.backward()
```

```python
gradient = input_var.grad
gradient
```

```python
input_var.grad = None
```

```python
import torch
time = torch.tensor(0.11, requires_grad=True)
position = 4.9 * time**2
position.backward(create_graph=True)
speed = time.grad
time.grad = None
speed.backward()
acceleration = time.grad
print(f"Time:         {time.item():.4f}")
print(f"Position:     {position.item():.4f}")
print(f"Speed:        {speed.item():.4f}")
print(f"Acceleration: {acceleration.item():.4f}")
```

```python
import torch

time = torch.tensor(0.11, requires_grad=True)
position = 4.9 * time**2
speed = torch.autograd.grad(outputs=position, inputs=time, create_graph=True)[0]
acceleration = torch.autograd.grad(outputs=speed, inputs=time)[0]
print(f"Time:         {time.item():.4f}")
print(f"Position:     {position.item():.4f}")
print(f"Speed:        {speed.item():.4f}")
print(f"Acceleration: {acceleration.item():.4f}")
```
