```python
import numpy as np
```

```python
# scalar / rank-0 Tensor / 0D tensor
x = np.array(12)
print("Dim/Rank :", x.ndim)
print("Shape :", x.shape)
print("X :", x)
```

```python
# vector / rank-1 Tensor / 1D tensor
x = np.array([12, 3, 6, 14, 7])
print("Dim/Rank :", x.ndim)
print("Shape :", x.shape)
print("X :", x)
```

```python
# matrix / rank-2 Tensor \ 2D tensor
x = np.array([[5, 78, 2, 34, 0],[6, 79, 3, 35, 1],[7, 80, 4, 36, 2]])
print("Dim/Rank :", x.ndim)
print("Shape :", x.shape)
print("X :")
x
```

```python
# Rank-3 Tensor / 3D Tensor
x = np.array([[[5, 78, 2, 34, 0],
[6, 79, 3, 35, 1],
[7, 80, 4, 36, 2]],
[[5, 78, 2, 34, 0],
[6, 79, 3, 35, 1],
[7, 80, 4, 36, 2]],
[[5, 78, 2, 34, 0],
[6, 79, 3, 35, 1],
[7, 80, 4, 36, 2]]])

print("Dim/Rank :", x.ndim)
print("Shape :", x.shape)
print("X :")
x
```
