```python
import keras
import numpy as np
import matplotlib.pyplot as plt
```

IMBD DATASET

### Dataset

```python
# Fetch Dataset (IMBD)
# num_words indicate top max freq occurring words
num_words = 10000
(train_data, train_labels), (test_data, test_labels) = keras.datasets.imdb.load_data(num_words=num_words)
print(train_data.shape, train_labels.shape, test_data.shape, test_labels.shape)
```

```python
# Tokenizer mapping
word_index = keras.datasets.imdb.get_word_index()
word_index
```

```python
# Reversed Tokenizer mapping
reverse_word_index = dict([(value, key) for (key, value) in word_index.items()])
reverse_word_index
```

```python
# Decoding First Review
decoded_review = " ".join([reverse_word_index.get(i - 3, "?") for i in train_data[0]])
decoded_review
```

```python
# Encoding the integer sequences via multi-hot encoding
def multi_hot_encode(sequences, num_classes):
    results = np.zeros((len(sequences), num_classes))
    for i, sequence in enumerate(sequences):
        results[i][sequence] = 1.0
    return results


x_train = multi_hot_encode(train_data, num_classes=num_words)
x_test = multi_hot_encode(test_data, num_classes=num_words)
```

```python
x_train[0]
```

```python
x_test[0]
```

```python
y_train = train_labels.astype("float32")
y_test = test_labels.astype("float32")
y_train
```

### Modelling and Training

```python
# Create Model
model = keras.Sequential(
[
        keras.layers.Dense(16, activation="relu"),
        keras.layers.Dense(16, activation="relu"),
        keras.layers.Dense(1, activation="sigmoid"),
]
)
model.compiled
```

```python
# Compile Model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"],
)
model.compiled
```

```python
model
```

```python
# Create Vlaidation Set
x_val = x_train[:10000]
partial_x_train = x_train[10000:]
y_val = y_train[:10000]
partial_y_train = y_train[10000:]
```

```python
history = model.fit(
    partial_x_train,
    partial_y_train,
    epochs=20,
    batch_size=512,
    validation_data=(x_val, y_val),
)
```

```python
history.history.keys()
```

### Assessing

```python
history_dict = history.history
loss_values = history_dict["loss"]
val_loss_values = history_dict["val_loss"]
epochs = range(1, len(loss_values) + 1)
plt.plot(epochs, loss_values, "r--", label="Training loss")
plt.plot(epochs, val_loss_values, "b", label="Validation loss")
plt.title("[IMDB] Training and validation loss")
plt.xlabel("Epochs")
plt.xticks(epochs)
plt.ylabel("Loss")
plt.legend()
plt.show()
```

```python
plt.clf()
acc = history_dict["accuracy"]
val_acc = history_dict["val_accuracy"]
plt.plot(epochs, acc, "r--", label="Training acc")
plt.plot(epochs, val_acc, "b", label="Validation acc")
plt.title("[IMDB] Training and validation accuracy")
plt.xlabel("Epochs")
plt.xticks(epochs)
plt.ylabel("Accuracy")
plt.legend()
plt.show()
```

### Optimization

```python
model = keras.Sequential(
[
        keras.layers.Dense(16, activation="relu"),
        keras.layers.Dense(16, activation="relu"),
        keras.layers.Dense(1, activation="sigmoid"),
    ]
)
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"],
)
model.fit(x_train, y_train, epochs=4, batch_size=512)
```

```python
results = model.evaluate(x_test, y_test)
results
```

### Prediction

```python
y_pred = model.predict(x_test)
y_pred
```

### Model Evaluation

```python
y_pred_binary = (y_pred > 0.5).astype("int32")
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred_binary, target_names=['Negative', 'Positive']))
```
